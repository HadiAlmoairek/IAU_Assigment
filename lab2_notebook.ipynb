## 6. Generate & Save Methodology Diagram

```python
fig, ax = plt.subplots(figsize=(10, 15))
ax.set_xlim(0, 10)
ax.set_ylim(0, 16)
ax.axis('off')
fig.patch.set_facecolor('#F0F4F8')
ax.set_facecolor('#F0F4F8')

steps = [
    ('1. Data Collection',      '#4472C4', 'Load insurance.csv · 1338 rows × 7 cols'),
    ('2. EDA',                  '#ED7D31', 'Distributions · Correlations · Smoker vs Charges'),
    ('3. Preprocessing',        '#70AD47', 'Outlier removal (IQR) · Normalization · Encoding'),
    ('4. Feature Engineering',  '#FFC000', 'BMI category · Age group · Interaction features'),
    ('5. Train / Test Split',   '#9DC3E6', '80 % Train  |  20 % Test  (random_state=42)'),
    ('6. Model Training',       '#F4B183', 'Linear Regression · Logistic Regression · KNN'),
    ('7. Evaluation',           '#C5E0B4', 'MAE · RMSE (regression) · F1 · AUC (classification)'),
    ('8. Iteration & Tuning',   '#D9D9D9', 'Hyperparameter search · PCA · Feature selection'),
]

box_h, gap, start_y = 1.35, 0.35, 15.2

for i, (title, color, desc) in enumerate(steps):
    y = start_y - i * (box_h + gap)
    rect = plt.Rectangle((0.8, y - box_h), 8.4, box_h,
                          facecolor=color, edgecolor='#333', linewidth=1.5, zorder=2,
                          clip_on=False)
    ax.add_patch(rect)
    ax.text(5, y - box_h/2 + 0.22, title, ha='center', va='center',
            fontsize=11.5, fontweight='bold', color='#1a1a1a', zorder=3)
    ax.text(5, y - box_h/2 - 0.22, desc, ha='center', va='center',
            fontsize=9, color='#333', zorder=3)
    if i < len(steps) - 1:
        ay = y - box_h
        ax.annotate('', xy=(5, ay - gap), xytext=(5, ay),
                    arrowprops=dict(arrowstyle='->', color='#444', lw=2), zorder=4)

ax.set_title('ML Workflow — Medical Insurance Cost Prediction',
             fontsize=13, fontweight='bold', pad=12, color='#1a1a1a')
plt.tight_layout()
plt.savefig('methodology_diagram.png', dpi=150, bbox_inches='tight',
            facecolor=fig.get_facecolor())
plt.show()
print('Saved: methodology_diagram.png')
```
