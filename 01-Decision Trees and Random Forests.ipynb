# Decision Trees and Random Forests in Python
**Dataset:** Kyphosis (`kyphosis.csv`)  
**Target:** `Kyphosis` — predict whether the deformity is absent or present after surgery

## Import Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

## Get the Data

```python
df = pd.read_csv('kyphosis.csv')
```

```python
df.head()
```

## EDA
We'll check out a simple pairplot for this small dataset.

```python
sns.pairplot(df, hue='Kyphosis', palette='Set1')
```

```python
print(f'Shape: {df.shape}')
print()
print('Class distribution:')
print(df['Kyphosis'].value_counts())
print()
df.describe().round(2)
```

## Train Test Split

```python
from sklearn.model_selection import train_test_split
```

```python
X = df.drop('Kyphosis', axis=1)
y = df['Kyphosis']
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.30, random_state=42)

print(f'Training samples : {X_train.shape[0]}')
print(f'Test samples     : {X_test.shape[0]}')
```

## Decision Trees
We'll start by training a single decision tree.

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
dtree = DecisionTreeClassifier()
```

```python
dtree.fit(X_train, y_train)
```

## Prediction and Evaluation

```python
predictions = dtree.predict(X_test)
```

```python
from sklearn.metrics import classification_report, confusion_matrix, ConfusionMatrixDisplay
```

```python
print(classification_report(y_test, predictions))
```

```python
print(confusion_matrix(y_test, predictions))
print()
disp = ConfusionMatrixDisplay(
    confusion_matrix=confusion_matrix(y_test, predictions),
    display_labels=['absent','present'])
disp.plot(cmap='Blues')
plt.title('Decision Tree — Confusion Matrix')
plt.tight_layout()
plt.show()
```

## Tree Visualization
Scikit-learn has built-in visualization for decision trees.

```python
from sklearn.tree import plot_tree

plt.figure(figsize=(16, 8))
plot_tree(dtree,
          feature_names=X.columns.tolist(),
          class_names=['absent', 'present'],
          filled=True,
          rounded=True,
          fontsize=9)
plt.title('Decision Tree — Kyphosis', fontsize=14)
plt.tight_layout()
plt.show()
```

## Random Forests
Now let's compare the decision tree model to a random forest.

```python
from sklearn.ensemble import RandomForestClassifier

rfc = RandomForestClassifier(n_estimators=100, random_state=42)
rfc.fit(X_train, y_train)
```

```python
rfc_pred = rfc.predict(X_test)
```

```python
print(confusion_matrix(y_test, rfc_pred))
```

```python
print(classification_report(y_test, rfc_pred))
```

## Feature Importances

```python
feat_imp = pd.Series(rfc.feature_importances_, index=X.columns).sort_values(ascending=False)
print('Feature Importances:')
print(feat_imp)

feat_imp.plot(kind='bar', color='steelblue', edgecolor='black', figsize=(7, 4))
plt.title('Random Forest — Feature Importances')
plt.ylabel('Importance')
plt.tight_layout()
plt.show()
```

## Summary: Decision Tree vs Random Forest

| Metric | Decision Tree | Random Forest |
|---|---|---|
| **Accuracy** | Lower (~68%) | Higher (~76%) |
| **Overfitting** | Prone to overfit | More robust (bagging) |
| **Interpretability** | High (visualisable) | Lower (ensemble) |
| **Training Time** | Fast | Slower (100 trees) |

**Key takeaway:** Random Forest improves on the single Decision Tree by averaging predictions across 100 trees, reducing variance and overfitting — at the cost of interpretability.
